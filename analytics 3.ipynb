```python
#droping the column
X=df.drop(["species"],axis=1)
y=df["species"]
```

```python
X
```

```python
y
```

```python
#scale them in common range so we use min max scalar
from sklearn.preprocessing import MinMaxScaler
scalar=MinMaxScaler()
X_scaled=scalar.fit_transform(X)
X_scaled
```

```python
#create object of classifier
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,test_size=0.2,random_state=42)
```

```python
from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
gnb.fit(X_train,y_train)
y_pred=gnb.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
import sklearn.metrics
x=['Iris-setosa','Iris-virginica','Iris-versicolor']
print(sklearn.metrics.confusion_matrix(y_test,y_pred,labels=x))
# along the princiapl diagonal we have true positives 10,11,9 are TP,
#so no clasification error.will get 100 % accuracy
```

```python
import sklearn.metrics

lbs = ['Iris-versicolor', 'Iris-setosa', 'Iris-virginica']

print(sklearn.metrics.confusion_matrix(y_test, y_pred, labels=lbs))
print(sklearn.metrics.accuracy_score(y_test, y_pred))
print(sklearn.metrics.precision_score(y_test, y_pred, labels=lbs, pos_label=1, average='micro'))
print(sklearn.metrics.recall_score(y_test, y_pred, labels=lbs, pos_label=1, average='micro'))
print(sklearn.metrics.f1_score(y_test, y_pred, labels=lbs, pos_label=1, average='micro'))
```

```python
#Classfication Report
from sklearn.metrics import classification_report
R = classification_report(y_test, y_pred)
print(R)
```
